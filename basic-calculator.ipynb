```python
import math
```

```python
class Calculator:
    """A calculator that supports basic and advanced math operations,
    with the ability to register new operations dynamically."""

    def __init__(self):
        self.operations = {}
        self.init()

    def init(self):
        """Initialize the dictionary with the basic operations."""
        self.operations['+'] = lambda x, y: x + y
        self.operations['-'] = lambda x, y: x - y
        self.operations['*'] = lambda x, y: x * y
        self.operations['/'] = lambda x, y: x / y if y != 0 else self._div_by_zero()

    def _div_by_zero(self):
        print("Error: Division by zero is not allowed.")
        raise ZeroDivisionError("Division by zero")

    def add_operation(self, symbol, func):
        """Add a new operation and its function to the dictionary."""
        self.operations[symbol] = func

    def calculate(self, num1, symbol, num2):
        """Perform a calculation using the operation dictionary,
        with error handling for invalid operations and non-numeric input."""
        if symbol not in self.operations:
            print(f"Error: '{symbol}' is not a valid operation.")
            raise ValueError(f"Invalid operation: {symbol}")

        if not isinstance(num1, (int, float)) or not isinstance(num2, (int, float)):
            print("Error: Both inputs must be numbers.")
            raise TypeError("Inputs must be numbers")

        try:
            return self.operations[symbol](num1, num2)
        except ZeroDivisionError:
            raise
        except ValueError as e:
            print(f"Error: Invalid math operation - {e}")
            raise
        except Exception as e:
            print(f"Error: An unexpected error occurred - {e}")
            raise
```

```python
def exponentiate(x, y):
    """Return x raised to the power of y."""
    return math.pow(x, y)
```

```python
def square_root(x, y=None):
    """Return the square root of x. The second argument is ignored,
    but kept so the function matches the calculator's two-argument
    operation signature."""
    if x < 0:
        print("Error: Cannot take the square root of a negative number.")
        raise ValueError("Square root of negative number")
    return math.sqrt(x)
```

```python
def logarithm(x, y):
    """Return the logarithm of x with base y."""
    if x <= 0 or y <= 0 or y == 1:
        print("Error: Invalid values for logarithm (x must be > 0, base must be > 0 and != 1).")
        raise ValueError("Invalid logarithm arguments")
    return math.log(x, y)
```

```python
def get_number(prompt):
    """Prompt the user until they enter a valid number, then return it
    as an int or float."""
    while True:
        value = input(prompt)
        try:
            num = float(value)
            if num.is_integer():
                num = int(num)
            if isinstance(num, (int, float)):
                return num
        except ValueError:
            print("That's not a valid number. Please try again.")
```

```python
def main():
    calc = Calculator()

    # Register the advanced operations
    calc.add_operation('^', exponentiate)   # exponentiation
    calc.add_operation('sqrt', square_root)  # square root (second number ignored)
    calc.add_operation('log', logarithm)     # logarithm: log(x) base y

    print("=== Advanced Calculator ===")
    print("Available operations: + - * / ^ sqrt log")
    print("  sqrt example: enter first number, then any value for the second (it's ignored)")
    print("  log example: first number = value, second number = base")
    print("Type 'exit' at any prompt to quit.\n")

    while True:
        symbol = input("Enter operation (+, -, *, /, ^, sqrt, log) or 'exit' to quit: ").strip()

        if symbol.lower() == 'exit':
            print("Goodbye!")
            break

        if symbol not in calc.operations:
            print(f"'{symbol}' is not a recognized operation. Please try again.\n")
            continue

        num1 = get_number("Enter the first number: ")

        if symbol == 'sqrt':
            # Square root only needs one number, but calculate() expects two
            num2 = 0
        else:
            num2 = get_number("Enter the second number: ")

        try:
            result = calc.calculate(num1, symbol, num2)
            print(f"Result: {result}\n")
        except Exception:
            # Error message was already printed inside calculate()
            print("Calculation failed. Please try again.\n")
            continue

        again = input("Would you like to perform another calculation? (y/n): ").strip().lower()
        if again != 'y':
            print("Goodbye!")
            break


if __name__ == "__main__":
    main()
```
